```python
import numpy as np
import matplotlib.pyplot as plt

signal = np.array([1, 2, 3, 4, 5, 6, 7, 8])
n = np.arange(len(signal))

fft_result = np.fft.fft(signal)

magnitude_spectrum = np.abs(fft_result)
phase_spectrum = np.angle(fft_result)

reconstructed_signal = np.fft.ifft(fft_result)

print("Original Signal:", signal)
print("\nFFT Result (Complex Values):\n", np.round(fft_result, 4))
print("\nMagnitude Spectrum:\n", np.round(magnitude_spectrum, 4))
print("\nPhase Spectrum (radians):\n", np.round(phase_spectrum, 4))
print("\nReconstructed Signal (IFFT):\n", np.round(reconstructed_signal.real, 4))

plt.figure(figsize=(12, 8))

# Subplot 1: Original Signal
plt.subplot(2, 2, 1)
plt.stem(n, signal)
plt.title("Original Discrete Signal x[n]")
plt.xlabel("Sample (n)")
plt.ylabel("Amplitude")
plt.grid(True)

# Subplot 2: Magnitude Spectrum
plt.subplot(2, 2, 2)
plt.stem(n, magnitude_spectrum)
plt.title("Magnitude Spectrum |X(k)|")
plt.xlabel("Frequency Bin (k)")
plt.ylabel("Magnitude")
plt.grid(True)

# Subplot 3: Phase Spectrum
plt.subplot(2, 2, 3)
plt.stem(n, phase_spectrum)
plt.title("Phase Spectrum ∠X(k)")
plt.xlabel("Frequency Bin (k)")
plt.ylabel("Phase (radians)")
plt.grid(True)

# Subplot 4: Reconstructed Signal from IFFT
plt.subplot(2, 2, 4)
plt.stem(n, reconstructed_signal.real)
plt.title("Reconstructed Signal from IFFT")
plt.xlabel("Sample (n)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
